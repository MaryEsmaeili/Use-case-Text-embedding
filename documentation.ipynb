# Clustering text

### Comparision the provided notebook and the article

The provided notebook and the research article share similarities but differ in their scope and goals. Both of them emphasize cleaning the text by removing unnecessary elements like punctuation, numbers, and irrelevant words and spaces.
The notebook creates a DTM using TF-IDF, where each document is represented by the frequency of its terms(max_df=0.8, min_df=0.01). The article also uses a similar DTM to represent documents based on important features like words, symptoms, and medications.
The notebook applies NMF (Non-negative Matrix Factorization) to reduce the DTM into topics, identifying key terms for each topic (which further add LDA too). The article uses NMF to group clinical documents into clusters based on their content.
Both methods aim to group similar documents together, either into topics (notebook) or meaningful clusters like symptom-based or medication-based groups (article).

### What Needs to Be Added to the Notebook

To align the notebook with the article, several key enhancements are required for example the notebook focuses on extracting nouns from the text, but the article emphasizes symptom names (fever, hypertension) and medication names (aspirin,ibuprofen).
Tools like MetaMap and MedEx are suggested by the article to extract these domain-specific terms instead of relying solely on general nouns, you need to focus on extracting clinical terms explicitly related to symptoms and medications.

The notebook uses a single DTM, but the article builds three separate representations of the data, which are processed separately and then combined for a more comprehensive analysis.
1.	Words: General vocabulary from the text.
2.	Symptoms: Terms extracted specifically related to medical symptoms.
3.	Medications: Terms extracted specifically related to medications.

The notebook applies a simple NMF model, but the article uses multi-view NMF, which integrates multiple sources of information, this approach allows the clustering process to consider different aspect of the text simultaneously, leading to better groupings.

The notebook doesn't evaluate the clustering results, while the article uses specific metrics such as Accuracy (Measures how well the clusters match known categories) and Normalized Mutual Information (NMI) (Evaluates how much information the clustering results retain about the original categories) adding these metrics allow us to compare the quality of clustering results under different settings.

At the end, The notebook outputs the top words for each topic but lacks advanced visualizations but the article emphasizes comparing results under different conditions. Bar charts and cluster accuracy plots can visually demonstrate these comparisons.
